```python
# Assignment-1
n = int(input("Enter number of time steps: "))

rewards = []

for i in range(n):
    reward = float(input(f"Enter reward for time step {i + 1}: "))
    rewards.append(reward)

gamma = float(input("Enter discount factor (gamma): "))

undiscounted_return = sum(rewards)

discounted_return = 0

for i in range(n):
    discounted_return += rewards[i] * (gamma ** i)

print("\nEntered reward values:", rewards)
print("Undiscounted return:", undiscounted_return)
print("Discounted return:", discounted_return)
```

```python
%%writefile grid_env_det.py

class GridWorld:
    def __init__(self):
        self.states = ["S0", "S1", "S3", "S4", "G"]
        self.actions = ["up", "down", "left", "right"]

        self.rewards = {
            "S0": -1,
            "S1": -1,
            "S3": -1,
            "S4": -1,
            "G": 10
        }

        self.obstacle_reward = -5
        self.gamma = 0.9

        self.transitions = {
            "S0": {
                "up": "S0",
                "down": "S0",
                "left": "S0",
                "right": "S1"
            },
            "S1": {
                "up": "S1",
                "down": "S3",
                "left": "S0",
                "right": "G"
            },
            "S3": {
                "up": "S1",
                "down": "S3",
                "left": "S3",
                "right": "S4"
            },
            "S4": {
                "up": "G",
                "down": "S4",
                "left": "S3",
                "right": "S4"
            },
            "G": {
                "up": "G",
                "down": "G",
                "left": "G",
                "right": "G"
            }
        }

    def get_next_state(self, state, action):
        return self.transitions[state][action]

    def get_reward(self, state):
        return self.rewards.get(state, self.obstacle_reward)
```

```python
%%writefile main.py

from grid_env_det import GridWorld

env = GridWorld()

print("States:", env.states)
print("Actions:", env.actions)
print("Discount factor:", env.gamma)

state = input("Enter current state: ")
action = input("Enter action: ").lower()

if state not in env.states or action not in env.actions:
    print("Invalid state or action")
else:
    next_state = env.get_next_state(state, action)

    print("Current state:", state)
    print("Action:", action)
    print("Next state:", next_state)

    if next_state == "G":
        reward = 10
    else:
        reward = -1

    print("Reward:", reward)
```
