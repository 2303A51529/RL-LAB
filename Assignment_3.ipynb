```python
import random

machines = ['A', 'B', 'C']

N_play = 10
epsilon = 0.2

N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': 0, 'B': 0, 'C': 0}
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}

for play in range(1, N_play + 1):

    r = random.random()

    best_machine = max(Q, key=Q.get)

    if r < epsilon:
        other_machines = [
            m for m in machines if m != best_machine
        ]
        selected = random.choice(other_machines)
        print("Exploration")
    else:
        selected = best_machine
        print("Exploitation")

    print("Play:", play)
    print("Selected Machine:", selected)

    reward = float(input("Enter reward: "))

    N[selected] += 1
    R[selected] += reward

    Q[selected] = Q[selected] + (
        reward - Q[selected]
    ) / N[selected]

    print("N:", N)
    print("R:", R)
    print("Q:", {m: round(Q[m], 2) for m in machines})
    print("-" * 40)

print("\nFINAL RESULTS")
print("-" * 40)

print("Machine\tN\tR\tQ")

for machine in machines:
    print(
        machine,
        "\t", N[machine],
        "\t", R[machine],
        "\t", round(Q[machine], 2)
    )

total_reward = sum(R.values())

print("-" * 40)
print("Total Reward =", total_reward)5
```
