```python

gamma = 0.9
theta = 0.001

states = ["S0", "S1"]
actions = ["Up", "Down", "Left", "Right"]

transitions = {
    "S0": {
        "Up": ("S0", -1),
        "Down": ("S1", -1),
        "Left": ("S0", -1),
        "Right": ("S0", -1)
    },
    "S1": {
        "Up": ("S0", -1),
        "Down": ("S1", -1),
        "Left": ("S1", -1),
        "Right": ("G", 100)
    }
}

values = {
    "S0": 0.0,
    "S1": 0.0,
    "G": 0.0
}

print("STATE, ACTION AND TRANSITION DETAILS")

for state in states:
    for action in actions:
        next_state, reward = transitions[state][action]
        print(state, action, "->", next_state, "Reward:", reward)

print("\nINITIAL STATE VALUES")
print(values)

print("\nVALUE ITERATION")

iteration = 0

while True:
    new_values = values.copy()

    for state in states:
        q_values = []

        for action in actions:
            next_state, reward = transitions[state][action]

            q = reward + gamma * values[next_state]
            q_values.append(q)

        new_values[state] = max(q_values)

    delta = max(
        abs(new_values[state] - values[state])
        for state in states
    )

    iteration += 1
    values = new_values

    print(
        f"Iteration {iteration}: "
        f"S0 = {values['S0']:.4f}, "
        f"S1 = {values['S1']:.4f}, "
        f"Delta = {delta:.4f}"
    )

    if delta < theta:
        break

print("\nFINAL OPTIMAL STATE VALUES")

for state in states:
    print(f"V*({state}) = {values[state]:.4f}")

print(f"V*(G) = {values['G']:.4f}")

print("\nFINAL ACTION VALUE TABLE")

q_table = {}

for state in states:
    q_table[state] = {}

    for action in actions:
        next_state, reward = transitions[state][action]

        q = reward + gamma * values[next_state]
        q_table[state][action] = q

        print(f"Q*({state}, {action}) = {q:.4f}")

print("\nOPTIMAL POLICY")

policy = {}

for state in states:
    best_value = max(q_table[state].values())

    best_actions = [
        action
        for action in actions
        if abs(q_table[state][action] - best_value) < 1e-9
    ]

    policy[state] = best_actions

    print(f"{state} = {best_actions}")

print("\nBELLMAN OPTIMALITY VERIFICATION")

for state in states:
    max_q = max(q_table[state].values())

    verified = abs(values[state] - max_q) < 1e-9

    print(
        state,
        "V* =", values[state],
        "Max Q* =", max_q,
        "Verified =", verified
    )

print("\nOPTIMAL PATH: S0 -> S1 -> G")
```
