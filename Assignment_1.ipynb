```python
# problem-1
n = int(input("Enter the number of machines: "))
machines = {}
for i in range(n):
    name = input(f"\nEnter name of machine {i + 1}: ")
    rewards = list(map(
        float,
        input(f"Enter rewards for Machine {name} separated by spaces: ").split()
    ))
    machines[name] = rewards
q_values = {}
print("\n--- Machine Details ---")
for name, rewards in machines.items():
    q = sum(rewards) / len(rewards)
    q_values[name] = q
    print(f"\nMachine {name}")
    print("Rewards:", rewards)
    print(f"Q({name}) = {q:.2f}")
selected_machine = max(q_values, key=q_values.get)
print("\n------------------------")
print("Greedy Selected Machine:", selected_machine)
print(f"Highest Q value = {q_values[selected_machine]:.2f}")
```

```python
# Problem-2
n = int(input("Enter the number of machines: "))
machines = {}
for i in range(n):
    name = input(f"\nEnter name of machine {i + 1}: ")
    rewards = list(map(
        float,
        input(f"Enter rewards for Machine {name} separated by spaces: ").split()
    ))
    machines[name] = rewards
q_values = {}
counts = {}
for name in machines:
    q_values[name] = 0
    counts[name] = 0
print("\n--- Incremental Updates ---")
for name, rewards in machines.items():
    print(f"\nMachine {name}")
    for reward in rewards:
        counts[name] += 1
        old_q = q_values[name]
        q_values[name] = old_q + (
            reward - old_q
        ) / counts[name]
        print(
            f"Reward = {reward:.2f}, "
            f"N = {counts[name]}, "
            f"Q({name}) = {q_values[name]:.2f}"
        )
print("\n--- Final Q Values ---")
for name in q_values:
    print(f"Q({name}) = {q_values[name]:.2f}")
selected_machine = max(q_values, key=q_values.get)
print("\n------------------------")
print("Greedy Selected Machine:", selected_machine)
print(f"Highest Q value = {q_values[selected_machine]:.2f}")
```
