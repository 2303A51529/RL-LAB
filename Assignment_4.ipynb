```python
import math

machines = ['A', 'B', 'C']

N_play = 10

N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': 0, 'B': 0, 'C': 0}
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}
UCB = {'A': 0.0, 'B': 0.0, 'C': 0.0}

for play in range(1, N_play + 1):

    for machine in machines:
        if N[machine] == 0:
            UCB[machine] = float('inf')
        else:
            UCB[machine] = Q[machine] + math.sqrt(
                (2 * math.log(play)) / N[machine]
            )

    selected = max(UCB, key=UCB.get)

    print("\nPlay:", play)
    print("Selected Machine:", selected)

    reward = float(input("Enter reward: "))

    N[selected] += 1
    R[selected] += reward

    Q[selected] = Q[selected] + (
        reward - Q[selected]
    ) / N[selected]

    print("N:", N)
    print("R:", R)
    print("UCB:", {
        m: round(UCB[m], 2) if UCB[m] != float('inf') else "inf"
        for m in machines
    })
    print("Q:", {
        m: round(Q[m], 2)
        for m in machines
    })

print("\nFINAL RESULTS")
print("-" * 45)
print("Machine\tN\tR\tUCB\tQ")

for machine in machines:
    print(
        machine,
        "\t", N[machine],
        "\t", R[machine],
        "\t", round(UCB[machine], 2)
        if UCB[machine] != float('inf') else "inf",
        "\t", round(Q[machine], 2)
    )

total_reward = sum(R.values())

print("-" * 45)
print("Total Reward =", total_reward)
```
